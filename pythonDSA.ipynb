# singly linked list

```python
class SinglyNode:
  def __init__(self,val,next=None):
    self.val = val
    self.next = next

  def __str__(self):
    return str(self.val)
```

```python
Head = SinglyNode(1)
A = SinglyNode(3)
B = SinglyNode(4)
C = SinglyNode(4)

Head.next = A
A.next = B
B.next = C
```

```python
# traversing
curr = Head
while curr:
  print(curr)
  curr = curr.next
```

```python
#display the linked list

def display(head):
  curr = head
  elements = []
  while curr:
    elements.append(str(curr.val))
    curr = curr.next
  print(" -> ".join(elements))
```

```python
display(Head)
```

```python
# search for a val

def search(head,val):
  curr = head
  while curr:
    if curr.val == val:
      return True
    curr = curr.next
  return False

print(search(Head,4),search(Head,9))
```

```python
class DoublyNode:
  def __init__(self,val,prev=None,next=None):
    self.val = val
    self.prev = prev
    self.next = next

  def __str__(self):
    return str(self.val)
```

```python
head = tail = DoublyNode(1)
print(head,tail)
```

```python
def display(head):
  curr = head
  elements= []
  while curr:
    elements.append(str(curr.val))
    curr = curr.next
  print('<->'.join(elements))

display(head)
```

# recursion

```python
#fibo

def fibo(n):
  if n == 0:
    return 0
  if n == 1:
    return 1
  return fibo(n-1) + fibo(n-2)

fibo(6)
```

# binary search

```python
def binary_search(arr,target):

  if len(arr) == 0:
    return -1

  low = 0
  high = len(arr) - 1

  while(low <= high):
    mid = (low + high) // 2

    if arr[mid] == target:
      return mid
    elif arr[mid] > target:
      high = mid - 1
    else:
      low = mid + 1

  return -1
```

```python
arr = [1,3,55,67,889,12930]

print(binary_search(arr,12930))
```

```python
#over under condition based bs
arr = [False,False,False,True,True,True,True]

def BS(arr):
  left = 0
  right = len(arr) - 1

  while(left < right):
    mid = (left+right)//2

    if arr[mid]:
      right = mid
    else:
      left = mid + 1

  return left

BS(arr)
```

# Binary Tree and Binary Search Tree

```python
#binary tree

class TreeNode:
  def __init__(self,val,left=None,right=None):
    self.val = val
    self.left = left
    self.right = right

  def __str__(self):
    return str(self.val)
```

```python
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)

A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
```

```python
print(A)
```

```python
def pre_order(node):
  if node is None:
    return

  print(node.val)
  pre_order(node.left)
  pre_order(node.right)

pre_order(A)
```

```python
def in_order(node):
  if node is None:
    return

  in_order(node.left)
  print(node.val)
  in_order(node.right)

in_order(A)
```

```python
# bfs

from collections import deque

def bfs(node):
  q = deque()
  q.append(node)

  while(q):
    node = q.popleft()
    print(node)
    if node.left : q.append(node.left)
    if node.right : q.append(node.right)

bfs(A)
```

```python
#search dfs

def searchDFS(node,target):
  if node is None:
    return False

  if node.val == target:
    return True

  return searchDFS(node.left,target) or searchDFS(node.right,target)

searchDFS(A,19)
```

```python
# bst

A2 = TreeNode(5)
B2 = TreeNode(1)
C2 = TreeNode(8)
D2 = TreeNode(-1)
E2 = TreeNode(3)
F2 = TreeNode(7)
G2 = TreeNode(9)

A2.left, A2.right = B2, C2
B2.left, B2.right = D2, E2
C2.left, C2.right = F2, G2

in_order(A2)
```

```python
def search_BST(node,target):
  if node is None:
    return False

  if node.val == target: return True

  if node.val > target:
    return search_BST(node.left,target)
  else:
    return search_BST(node.right,target)

print(search_BST(A2,5))
print(search_BST(A2,33))
```
